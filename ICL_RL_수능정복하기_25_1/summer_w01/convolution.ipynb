```python
import numpy as np
import torch
import torch.nn.functional as F
import matplotlib.pyplot as plt
```

```python
x = torch.linspace(-0.5, 0.5, 100)
inputs = torch.ones(100)
# filter = 5/4 - (1/2 - torch.linspace(0, 1, 100)).abs()
filter = torch.relu(torch.linspace(-0.5, 0.5, 100)) * 8/5 + 4/5

plt.ylim(0, 2)
plt.plot(x, inputs, label="PDF")
plt.plot(x, filter, label="filter")
plt.legend()
```

```python
plt.ylim(0, 2)
plt.title("n=0.25")
plt.plot(x, inputs, label="PDF")
plt.plot(0.25-x, filter, label="filter")
```

```python
plt.ylim(0, 2)
plt.title("n=0.50")
plt.plot(x, inputs, label="PDF")
plt.plot(0.5-x, filter, label="filter")
```

```python
x = torch.linspace(-1, 1, 199)
c1nd = np.convolve(inputs.numpy(), inputs.numpy()) # (-1.0, 1.0)
c1nd /= c1nd.sum() * (1+1)/199

plt.plot(x, c1nd, label="PDF1nd")
plt.legend()
```

```python
np.convolve(inputs.numpy(), filter.numpy()).shape # (-1.0, 1.0)
```

```python
d = 0.5
cnd = inputs.numpy()
for i in range(100):
    d += 0.5
    cnd = np.convolve(cnd, inputs.numpy()) # (-1.0, 1.0)
    cnd /= cnd.sum() * 2*d/cnd.shape[0]
x = torch.linspace(-1, 1, cnd.shape[0])
plt.plot(x, cnd, label="PDF100nd")
plt.legend()
```
